```python
# ============================================
# COMPANY EMPLOYEE DATA CLEANING PROJECT
# ============================================

import pandas as pd

# --------------------------------------------
# 1. LOAD THE DATASET
# --------------------------------------------

# Upload your CSV file to Google Colab first.
# Then replace the filename below with your actual filename.

df = pd.read_csv("company_employee_dataset.csv")


# --------------------------------------------
# 2. INSPECT THE DATASET
# --------------------------------------------

print("First 5 rows:")
print(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()


# --------------------------------------------
# 3. CHECK MISSING VALUES
# --------------------------------------------

print("\nMissing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())


# --------------------------------------------
# 4. CHECK DUPLICATE RECORDS
# --------------------------------------------

print("\nNumber of duplicate records:")
print(df.duplicated().sum())


# --------------------------------------------
# 5. CHECK UNIQUE VALUES
# --------------------------------------------

print("\nDepartment values:")
print(df["Department"].unique())

print("\nGender values:")
print(df["Gender"].unique())

print("\nCity values:")
print(df["City"].unique())

print("\nWork Mode values:")
print(df["Work_Mode"].unique())


# --------------------------------------------
# 6. CONVERT N/A VALUES TO MISSING VALUES
# --------------------------------------------

# Some datasets use N/A instead of a blank value.
# Convert N/A into proper missing values.

df = df.replace(["N/A", "NA", "n/a", "na", ""], pd.NA)


# --------------------------------------------
# 7. STANDARDIZE TEXT VALUES
# --------------------------------------------

# Remove extra spaces and make department names consistent.

df["Department"] = (
    df["Department"]
    .astype("string")
    .str.strip()
    .str.title()
)

# Standardize Gender
df["Gender"] = (
    df["Gender"]
    .astype("string")
    .str.strip()
    .str.title()
)

# Standardize City
df["City"] = (
    df["City"]
    .astype("string")
    .str.strip()
    .str.title()
)

# Standardize Work Mode
df["Work_Mode"] = (
    df["Work_Mode"]
    .astype("string")
    .str.strip()
    .str.title()
)


# --------------------------------------------
# 8. CORRECT DATA TYPES
# --------------------------------------------

# Convert numerical columns to numeric type.

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

df["Annual_Salary"] = pd.to_numeric(
    df["Annual_Salary"],
    errors="coerce"
)

df["Experience_Years"] = pd.to_numeric(
    df["Experience_Years"],
    errors="coerce"
)

df["Performance_Score"] = pd.to_numeric(
    df["Performance_Score"],
    errors="coerce"
)

# Convert Joining_Date to datetime

df["Joining_Date"] = pd.to_datetime(
    df["Joining_Date"],
    errors="coerce"
)


# --------------------------------------------
# 9. HANDLE NUMERICAL MISSING VALUES
# --------------------------------------------

# Use MEDIAN for numerical columns.
# Median is useful because it is less affected by extreme values.

df["Age"] = df["Age"].fillna(df["Age"].median())

df["Annual_Salary"] = df["Annual_Salary"].fillna(
    df["Annual_Salary"].median()
)

df["Experience_Years"] = df["Experience_Years"].fillna(
    df["Experience_Years"].median()
)

df["Performance_Score"] = df["Performance_Score"].fillna(
    df["Performance_Score"].median()
)


# --------------------------------------------
# 10. HANDLE CATEGORICAL MISSING VALUES
# --------------------------------------------

# Use MODE for categorical columns.
# Mode = most frequently occurring value.

df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)

df["Gender"] = df["Gender"].fillna(
    df["Gender"].mode()[0]
)

df["City"] = df["City"].fillna(
    df["City"].mode()[0]
)

df["Work_Mode"] = df["Work_Mode"].fillna(
    df["Work_Mode"].mode()[0]
)

df["Employee_Name"] = df["Employee_Name"].fillna(
    df["Employee_Name"].mode()[0]
)

df["Job_Title"] = df["Job_Title"].fillna(
    df["Job_Title"].mode()[0]
)


# --------------------------------------------
# 11. HANDLE MISSING JOINING DATES
# --------------------------------------------

# Forward fill carries the previous available
# date into the missing position.

df["Joining_Date"] = df["Joining_Date"].ffill()


# --------------------------------------------
# 12. REMOVE DUPLICATE RECORDS
# --------------------------------------------

print("\nDuplicates before removing:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after removing:")
print(df.duplicated().sum())


# --------------------------------------------
# 13. FINAL CHECK
# --------------------------------------------

print("\nFinal dataset shape:")
print(df.shape)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nTotal missing values after cleaning:")
print(df.isnull().sum().sum())

print("\nDuplicate records after cleaning:")
print(df.duplicated().sum())

print("\nFinal data types:")
print(df.dtypes)

print("\nFinal cleaned dataset:")
print(df.head())


# --------------------------------------------
# 14. EXPORT CLEANED DATASET
# --------------------------------------------

df.to_csv(
    "Cleaned_Company_Employee_Dataset.csv",
    index=False
)

print("\nCleaned dataset exported successfully!")
```
