```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("/content/Student_Wellbeing_Survey.csv")

print(df.shape)
print(df.head())
print(df.isnull().sum())
```

```python
columns = ['Weekly_Study_Hours', 'Average_Sleep_Hours', 'Daily_Screen_Time_Hours',
           'Stress_Score', 'Academic_Readiness_Score']

for col in columns:
    mean_val = df[col].mean()
    median_val = df[col].median()
    mode_val = df[col].mode()[0]
    print(col)
    print("Mean:", round(mean_val, 2))
    print("Median:", round(median_val, 2))
    print("Mode:", round(mode_val, 2))
    print()
```

```python
for col in columns:
    data = df[col]

    range_val = data.max() - data.min()
    variance = data.var()
    std_dev = data.std()
    q1 = data.quantile(0.25)
    q3 = data.quantile(0.75)
    iqr = q3 - q1

    print(col)
    print("Range:", round(range_val, 2))
    print("Variance:", round(variance, 2))
    print("Std Dev:", round(std_dev, 2))
    print("Q1:", round(q1, 2))
    print("Q3:", round(q3, 2))
    print("IQR:", round(iqr, 2))
    print()

# Which has the greatest variability?
print("Academic_Readiness_Score has the highest standard deviation")
print("(largest spread of raw values around the mean)")
```

```python
outlier_columns = ['Weekly_Study_Hours', 'Daily_Screen_Time_Hours',
                    'Commute_Time_Minutes', 'Monthly_Discretionary_Spending']

for col in outlier_columns:
    data = df[col]
    q1 = data.quantile(0.25)
    q3 = data.quantile(0.75)
    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outliers = df[(data < lower_limit) | (data > upper_limit)]

    print(col)
    print("Lower limit:", round(lower_limit, 2))
    print("Upper limit:", round(upper_limit, 2))
    print("Number of outliers:", len(outliers))
    print()
```

```python
col = 'Monthly_Discretionary_Spending'
data = df[col]

q1 = data.quantile(0.25)
q3 = data.quantile(0.75)
iqr = q3 - q1
lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

print("BEFORE removing outliers")
print("Mean:", round(data.mean(), 2))
print("Median:", round(data.median(), 2))

clean_data = data[(data >= lower_limit) & (data <= upper_limit)]

print("\nAFTER removing outliers")
print("Mean:", round(clean_data.mean(), 2))
print("Median:", round(clean_data.median(), 2))
```

```python
total_students = len(df)

# Define events
event_A = df['Part_Time_Job'] == 'Yes'
event_B = df['Stress_Score'] >= 7
event_C = df['Scholarship'] == 'Yes'
event_D = df['Exercise_Days_Per_Week'] >= 3

# Individual probabilities
P_A = event_A.sum() / total_students
P_B = event_B.sum() / total_students
P_C = event_C.sum() / total_students
P_D = event_D.sum() / total_students

print("P(A):", round(P_A, 4))
print("P(B):", round(P_B, 4))
print("P(C):", round(P_C, 4))
print("P(D):", round(P_D, 4))

# Combined probabilities
A_and_B = (event_A & event_B).sum() / total_students
A_or_B = P_A + P_B - A_and_B

print("\nP(A and B):", round(A_and_B, 4))
print("P(A or B):", round(A_or_B, 4))

# Conditional probabilities
P_A_given_B = A_and_B / P_B
P_B_given_A = A_and_B / P_A

print("\nP(A|B):", round(P_A_given_B, 4))
print("P(B|A):", round(P_B_given_A, 4))
```

```python
year1 = df['Year_of_Study'] == 1
year4 = df['Year_of_Study'] == 4

overlap = (year1 & year4).sum()
print("Students who are both Year 1 and Year 4:", overlap)
print("Mutually exclusive?", overlap == 0)
```

```python
expected_if_independent = P_A * P_B

print("P(A and B) actual:", round(A_and_B, 4))
print("P(A) x P(B):", round(expected_if_independent, 4))

if abs(A_and_B - expected_if_independent) < 0.01:
    print("A and B look independent")
else:
    print("A and B are NOT independent")
```

```python
event_notA = df['Part_Time_Job'] == 'No'

P_B_given_notA = (event_B & event_notA).sum() / event_notA.sum()

# Bayes' formula
numerator = P_B_given_A * P_A
denominator = (P_B_given_A * P_A) + (P_B_given_notA * (1 - P_A))
P_A_given_B_bayes = numerator / denominator

print("P(B|not A):", round(P_B_given_notA, 4))
print("P(A|B) using Bayes:", round(P_A_given_B_bayes, 4))
print("P(A|B) direct calculation:", round(P_A_given_B, 4))
```

```python
scores = df['Academic_Readiness_Score']
mean_score = scores.mean()
std_score = scores.std()

print("Mean:", round(mean_score, 2))
print("Std Dev:", round(std_score, 2))

max_score = scores.max()
min_score = scores.min()

z_max = (max_score - mean_score) / std_score
z_min = (min_score - mean_score) / std_score

print("\nHighest score:", max_score, "-> Z-score:", round(z_max, 2))
print("Lowest score:", min_score, "-> Z-score:", round(z_min, 2))
```

```python
within_1sd = df[(scores >= mean_score - std_score) & (scores <= mean_score + std_score)]
within_2sd = df[(scores >= mean_score - 2*std_score) & (scores <= mean_score + 2*std_score)]
within_3sd = df[(scores >= mean_score - 3*std_score) & (scores <= mean_score + 3*std_score)]

pct_1sd = len(within_1sd) / total_students * 100
pct_2sd = len(within_2sd) / total_students * 100
pct_3sd = len(within_3sd) / total_students * 100

print("Within 1 SD:", round(pct_1sd, 1), "% (expected 68%)")
print("Within 2 SD:", round(pct_2sd, 1), "% (expected 95%)")
print("Within 3 SD:", round(pct_3sd, 1), "% (expected 99.7%)")
```
