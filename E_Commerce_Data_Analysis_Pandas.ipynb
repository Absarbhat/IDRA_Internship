# E-Commerce Sales Data Analysis using Python and Pandas

**Objective:** Analyze the provided e-commerce sales dataset using Pandas. The notebook covers data selection, filtering, sorting, `groupby()`, aggregation, sales statistics, and meaningful business observations.


## 1. Import Required Libraries

We use Pandas for data manipulation and analysis.

```python
import pandas as pd
import numpy as np




# Load data into DataFrame
df = pd.read_csv("/content/E_commerece_dataset.csv")

# Convert date column to datetime
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

print("Dataset loaded successfully!")
print("Shape:", df.shape)

```

## 4. Basic Data Selection

Select individual columns and multiple columns using Pandas.

```python
# Select one column
df['Product'].head()
```

```python
# Select multiple columns
df[['Order_ID', 'Customer', 'Product', 'Total_Sales']].head(10)
```

```python
# Select rows using loc
df.loc[0:4, ['Order_ID', 'City', 'Category', 'Total_Sales']]
```

## 5. Filtering Data

```python
# Orders with sales greater than 5,000
high_sales_orders = df[df['Total_Sales'] > 5000]
high_sales_orders[['Order_ID', 'Customer', 'Product', 'Total_Sales']].head(10)
```

```python
# Electronics orders
electronics_orders = df[df['Category'] == 'Electronics']
electronics_orders[['Order_ID', 'City', 'Product', 'Total_Sales']].head(10)
```

```python
# Orders from the North region with sales above 5,000
north_high_sales = df[(df['Region'] == 'North') & (df['Total_Sales'] > 5000)]
north_high_sales[['Order_ID', 'City', 'Category', 'Product', 'Total_Sales']]
```

## 6. Sorting Operations

```python
# Top 10 orders by total sales
top_orders = df.sort_values('Total_Sales', ascending=False).head(10)
top_orders[['Order_ID', 'Customer', 'City', 'Product', 'Total_Sales']]
```

```python
# Lowest 10 orders by total sales
lowest_orders = df.sort_values('Total_Sales', ascending=True).head(10)
lowest_orders[['Order_ID', 'Customer', 'Product', 'Total_Sales']]
```

## 7. Overall Sales Statistics

```python
total_sales = df['Total_Sales'].sum()
average_sales = df['Total_Sales'].mean()
highest_sales = df['Total_Sales'].max()
lowest_sales = df['Total_Sales'].min()
total_quantity = df['Quantity'].sum()
number_of_orders = df['Order_ID'].nunique()

print(f"Total Sales       : ₹{total_sales:,.2f}")
print(f"Average Sales     : ₹{average_sales:,.2f}")
print(f"Highest Sale      : ₹{highest_sales:,.2f}")
print(f"Lowest Sale       : ₹{lowest_sales:,.2f}")
print(f"Total Quantity    : {total_quantity:,}")
print(f"Number of Orders  : {number_of_orders:,}")
```

## 8. Category-wise Analysis

Use `groupby()` and aggregation to compare categories.

```python
category_analysis = df.groupby('Category').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Total_Quantity=('Quantity', 'sum'),
    Number_of_Orders=('Order_ID', 'nunique'),
    Average_Rating=('Rating', 'mean')
).sort_values('Total_Sales', ascending=False)

category_analysis
```

## 9. City-wise Analysis

```python
city_analysis = df.groupby('City').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Total_Quantity=('Quantity', 'sum'),
    Number_of_Orders=('Order_ID', 'nunique')
).sort_values('Total_Sales', ascending=False)

city_analysis
```

## 10. Product-wise Analysis

```python
product_analysis = df.groupby('Product').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Total_Quantity=('Quantity', 'sum'),
    Number_of_Orders=('Order_ID', 'nunique'),
    Average_Rating=('Rating', 'mean')
).sort_values('Total_Sales', ascending=False)

product_analysis
```

```python
# Top 5 products by total sales
product_analysis.head(5)
```

## 11. Payment Method Analysis

```python
payment_analysis = df.groupby('Payment_Method').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Number_of_Orders=('Order_ID', 'nunique')
).sort_values('Total_Sales', ascending=False)

payment_analysis
```

## 12. Region-wise Analysis

```python
region_analysis = df.groupby('Region').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Total_Quantity=('Quantity', 'sum'),
    Number_of_Orders=('Order_ID', 'nunique')
).sort_values('Total_Sales', ascending=False)

region_analysis
```

## 13. Identify Top-Performing Orders and Products

```python
# Highest-selling order
highest_order = df.loc[df['Total_Sales'].idxmax()]
highest_order[['Order_ID', 'Customer', 'City', 'Product', 'Quantity', 'Total_Sales']]
```

```python
# Lowest-selling order
lowest_order = df.loc[df['Total_Sales'].idxmin()]
lowest_order[['Order_ID', 'Customer', 'City', 'Product', 'Quantity', 'Total_Sales']]
```

```python
# Products with the highest quantity sold
top_quantity_products = (
    df.groupby('Product')['Quantity']
      .sum()
      .sort_values(ascending=False)
      .head(5)
)
top_quantity_products
```

## 14. Orders with High Ratings

```python
# Orders with a rating of 4.8 or higher
high_rated_orders = df[df['Rating'] >= 4.8].sort_values('Rating', ascending=False)
high_rated_orders[['Order_ID', 'Customer', 'Product', 'Rating', 'Total_Sales']].head(15)
```

## 15. Discount Analysis

```python
discount_analysis = df.groupby('Discount_Percent').agg(
    Total_Sales=('Total_Sales', 'sum'),
    Average_Sales=('Total_Sales', 'mean'),
    Number_of_Orders=('Order_ID', 'nunique')
).sort_values('Total_Sales', ascending=False)

discount_analysis
```

## 16. Meaningful Observations

The following cells automatically generate observations from the dataset, so the notebook remains accurate even if the CSV values are changed.

```python
top_category = category_analysis['Total_Sales'].idxmax()
top_category_sales = category_analysis.loc[top_category, 'Total_Sales']

top_city = city_analysis['Total_Sales'].idxmax()
top_city_sales = city_analysis.loc[top_city, 'Total_Sales']

top_product = product_analysis['Total_Sales'].idxmax()
top_product_sales = product_analysis.loc[top_product, 'Total_Sales']

top_payment = payment_analysis['Total_Sales'].idxmax()
top_payment_sales = payment_analysis.loc[top_payment, 'Total_Sales']

top_region = region_analysis['Total_Sales'].idxmax()
top_region_sales = region_analysis.loc[top_region, 'Total_Sales']

print("Key Observations")
print("-" * 60)
print(f"1. {top_category} is the highest-selling category with total sales of ₹{top_category_sales:,.2f}.")
print(f"2. {top_city} has the highest total sales among cities, with ₹{top_city_sales:,.2f}.")
print(f"3. {top_product} is the top-performing product by total sales, generating ₹{top_product_sales:,.2f}.")
print(f"4. {top_payment} is the payment method associated with the highest total sales, at ₹{top_payment_sales:,.2f}.")
print(f"5. {top_region} is the strongest region by total sales, contributing ₹{top_region_sales:,.2f}.")
print(f"6. The dataset contains {number_of_orders} orders and {total_quantity:,} units sold.")
print(f"7. The average order sales value is ₹{average_sales:,.2f}.")
```
