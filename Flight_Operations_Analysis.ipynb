```python
import pandas as pd
import numpy as np
df = pd.read_csv("/content/Flight_Operations_Dataset (1).csv")
```

**3.Data Cleaning**

```python
df.drop_duplicates(inplace=True)

df.isnull().sum()
```

**4.**Data** Selection**

```python
df[['Airline','Flight_ID']].head()
df.loc[:,['Flight_ID','Airline']]
df.iloc[:10]
```

**5.Data Filtering**

```python
df[df['Delay_Minutes'] > 30]

df[df['Flight_Status'] == 'Delayed']

df[(df['Delay_Minutes'] > 30) & (df['Flight_Status'] == 'Delayed')]
```

```python
df.sort_values(by='Delay_Minutes', ascending=False)

```

```python
df.groupby('Airline').agg({
    'Delay_Minutes':['mean','max','min'],
    'Flight_ID':'count'
})
```

```python
import numpy as np
df["Delay_category"]= np.where(df["Delay_Minutes"]>30,'high delay','low delay')
df.head()
```

```python
print("===== FLIGHT OPERATIONS OBSERVATIONS =====\n")

# 1
print("1. Dataset size:")
print(f"The dataset contains {len(df)} flights and {len(df.columns)} columns.\n")

# 2
airline_counts = df['Airline'].value_counts()
print("2. Airline with most flights:")
print(f"{airline_counts.index[0]} operates the highest number of flights "
      f"({airline_counts.iloc[0]} flights).\n")

# 3
avg_delay = df.groupby('Airline')['Delay_Minutes'].mean().sort_values(ascending=False)
print("3. Highest average delay:")
print(f"{avg_delay.index[0]} has the highest average delay "
      f"of {avg_delay.iloc[0]:.2f} minutes.\n")

# 4
print("4. Maximum delay:")
print(f"The maximum recorded delay is {df['Delay_Minutes'].max()} minutes.\n")

# 5
delayed = (df['Flight_Status'] == 'Delayed').sum()
print("5. Delayed flights:")
print(f"There are {delayed} delayed flights in the dataset.\n")

# 6
percentage = (delayed / len(df)) * 100
print("6. Percentage of delayed flights:")
print(f"{percentage:.2f}% of flights were delayed.\n")


# 7
destination_counts = df['Destination'].value_counts()
print("8. Busiest destination airport:")
print(f"{destination_counts.index[0]} receives the highest number of flights "
      f"({destination_counts.iloc[0]} flights).")
```
